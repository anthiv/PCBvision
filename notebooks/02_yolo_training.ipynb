# PCBVision – YOLO Baseline Training

## Objective

Train a baseline YOLO object detection model on the PCB defect dataset and evaluate whether the complete training pipeline works correctly.

This notebook will:

- verify the training environment
- load a pretrained YOLO model
- train it on the PCB dataset
- inspect training metrics
- evaluate it on the validation and test sets
- save the best model checkpoint

```python
from pathlib import Path

import torch
from ultralytics import YOLO

print(f"PyTorch version: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
print(f"Device: {'cuda' if torch.cuda.is_available() else 'cpu'}")
```

```python
PROJECT_ROOT = Path("..")
DATASET_PATH = PROJECT_ROOT / "dataset"
DATA_CONFIG_PATH = DATASET_PATH / "data.yaml"

assert DATA_CONFIG_PATH.exists(), f"Missing: {DATA_CONFIG_PATH}"
```

```python
model = YOLO("yolo11n.pt")
results = model.train(
    data=str(DATA_CONFIG_PATH),
    epochs=3,
    imgsz=640,
    batch=8,
    project=str(PROJECT_ROOT / "runs"),
    name="pcb_yolo11n_smoke_test",
)
```

```python
results = model.train(
    data=str(DATA_CONFIG_PATH),
    epochs=50,
    imgsz=640,
    batch=8,
    patience=10,
    project=str(PROJECT_ROOT / "runs"),
    name="pcb_yolo11n_baseline",
)
```
